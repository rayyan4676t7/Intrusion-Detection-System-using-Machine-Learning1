# Intrusion Detection System using Machine Learning

## Problem Statement
The goal is to classify network traffic as normal or attack using the NSL-KDD dataset.

```python
import pandas as pd

# Load dataset (make sure you uploaded files)
train = pd.read_csv("KDDTrain+.txt", header=None)
test = pd.read_csv("KDDTest+.txt", header=None)

train.head()
```

```python
columns = [
    "duration","protocol_type","service","flag","src_bytes","dst_bytes","land",
    "wrong_fragment","urgent","hot","num_failed_logins","logged_in",
    "num_compromised","root_shell","su_attempted","num_root","num_file_creations",
    "num_shells","num_access_files","num_outbound_cmds","is_host_login",
    "is_guest_login","count","srv_count","serror_rate","srv_serror_rate",
    "rerror_rate","srv_rerror_rate","same_srv_rate","diff_srv_rate",
    "srv_diff_host_rate","dst_host_count","dst_host_srv_count",
    "dst_host_same_srv_rate","dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate","dst_host_srv_diff_host_rate",
    "dst_host_serror_rate","dst_host_srv_serror_rate",
    "dst_host_rerror_rate","dst_host_srv_rerror_rate",
    "label","difficulty"
]

train.columns = columns
test.columns = columns

# Drop difficulty column
train = train.drop("difficulty", axis=1)
test = test.drop("difficulty", axis=1)

train.head()
```

```python
train['label'] = train['label'].astype(str)
test['label'] = test['label'].astype(str)

train['label'] = train['label'].apply(lambda x: 0 if 'normal' in x else 1)
test['label'] = test['label'].apply(lambda x: 0 if 'normal' in x else 1)

train['label'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

cols = ['protocol_type', 'service', 'flag']

for col in cols:
    train[col] = train[col].astype(str)
    test[col] = test[col].astype(str)

    le = LabelEncoder()
    combined = pd.concat([train[col], test[col]])
    le.fit(combined)

    train[col] = le.transform(train[col])
    test[col] = le.transform(test[col])

train.head()
```

```python
X_train = train.drop('label', axis=1)
y_train = train['label']

X_test = test.drop('label', axis=1)
y_test = test['label']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
import pandas as pd

X_train = pd.DataFrame(X_train)
X_test = pd.DataFrame(X_test)

X_train = X_train.dropna()
y_train = y_train[X_train.index]

X_test = X_test.dropna()
y_test = y_test[X_test.index]

X_train = X_train.values
X_test = X_test.values

# Check NaN
print(pd.DataFrame(X_train).isnull().sum().sum())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

model = LogisticRegression(max_iter=2000)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Logistic Regression:")
print(classification_report(y_test, y_pred))
```

```python
model_bal = LogisticRegression(max_iter=2000, class_weight={0:1, 1:2})
model_bal.fit(X_train, y_train)

y_pred_bal = model_bal.predict(X_test)

print("Balanced Logistic Regression:")
print(classification_report(y_test, y_pred_bal))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

print("Random Forest:")
print(classification_report(y_test, y_pred_rf))
```

## Results Interpretation

- Logistic Regression provides a strong baseline model.
- Balanced model improves recall (detects more attacks).
- Random Forest generally performs better due to non-linear learning.

### Key Insight:
In intrusion detection, **recall is more important than precision** because missing an attack is more dangerous than a false alarm.
